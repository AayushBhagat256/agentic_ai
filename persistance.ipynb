# Persistence 
<h4>Persistence in langGraph refer to the ability to save and restore the state of workflow over time</h4>

<br>

<p>Using Persistence you can actually save the values that are stored in the state so that in future you can actually restore those values</p>
<p>Does not stores just the final value of the state but it also stores the intermediatary values </p>
<p>Fault Tolerance Implementation - ability to resume from where it failed no need to re-run the entire thing</p>
<p>Applications : </p>
<ul>Chatbot - features like new coversation or continue some 3 days old chat </ul>

## Checkpointer in Persistence
<p> checkpoints -> states -> save  
<p> Every super step is a check point (a stage like start to next stage and so on is counted as single super set even parallel ones)
<p> persistence is implemented using checkpointer in langGraph

## Threads in persistence
<p> store values against a thread_id in db 
<p> retriving values against that thread _id will help us in resuming it
<p> Every time a workflow is executed a new thread_id is assigned
